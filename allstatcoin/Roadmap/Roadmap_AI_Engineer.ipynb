## Qu’est-ce qu’un(e) AI Engineer ?

Un(e) **AI Engineer** (ou ingénieur(e) en intelligence artificielle) est un professionnel chargé de **concevoir, implémenter, déployer et maintenir des systèmes IA complexes**, souvent construits autour de **models fondamentaux** (foundation models) comme les LLMs ou les modèles multimodaux.

Ce rôle ne se limite pas à l'entraînement de modèles. Il s’agit d’un métier **hybride** à la croisée :
- du machine learning,
- du développement logiciel,
- de l’ingénierie produit,
- et du MLOps.



Selon Chip Huyen (auteure et fondatrice de Claypot AI) :

> “**AI Engineering is not about building models. It's about building applications using foundation models.**”

Dans son ouvrage *AI Engineering: Building AI Applications with Foundation Models* (2024), Chip Huyen redéfinit l’AI Engineer comme :
- un **assembleur de briques d’IA** (pré-entraînées),
- capable de les **adapter**, **connecter**, **mettre en production**,
- et surtout de les **intégrer dans des produits réels**, robustes et utiles.

👉 Ce rôle est centré sur l’**impact concret**, l’**expérience utilisateur** et la **fiabilité à grande échelle**.


###  Compétences clés d’un AI Engineer :

- Maîtriser les **LLMs** et leurs APIs (OpenAI, Claude, Mistral…)
- Structurer du **Prompt Engineering** et du **Context Engineering**
- Créer des systèmes **RAG** (Retrieval-Augmented Generation)
- Développer des **agents IA autonomes**
- Exploiter les modèles **multimodaux** (texte, image, audio, code)
- Faire du **finetuning** (LoRA, QLoRA, PEFT)
- Déployer à l’aide de **Streamlit, FastAPI, Docker**
- Mettre en place du **monitoring**

#  Prompt Engineering & Context Engineering

Maîtriser le **prompt engineering** est une compétence clé pour exploiter toute la puissance des modèles de langage. C’est ce qui permet de transformer un LLM générique en un assistant ultra-performant, aligné sur une tâche, un ton ou un rôle spécifique.

Cette étape vous apprend à structurer vos requêtes, à concevoir des prompts clairs, robustes et orientés résultats, et à tirer parti des différentes stratégies de guidage et de formatage.

![prompt](img/prompt.png)

**L’objectif ici est d’acquérir :**
- Les grands types de prompting (zero-shot, few-shot, chain of thought)
- La capacité à définir des rôles clairs (persona prompting)
- La distinction entre **système prompt** et **prompt utilisateur**
- L’art de produire des réponses **structurées** (JSON, tableau, markdown…)
- L’utilisation de **frameworks** pour construire des prompts efficaces (CRAFTIF, TCRAEI, 5W, Meta-prompting)


##  Frameworks à utiliser
- **CRAFTIF** : Contexte, Rôle, Action, Format, Ton, Instruction, Feedback
- **TCRAEI** : Tâche, Contexte, Rôle, Audience, Exemple, Intention
- **5W** : Who, What, When, Where, Why – pour structurer les prompts complexes


## Fondamentaux du Prompting

- Générer automatiquement des prompts spécialisés (RH, juridique, SEO…)
- Créer des prompts qui construisent d'autres prompts
- Optimiser la formulation d’un prompt existant- Zero-shot, Few-shot, Chain of Thought (raisonnement explicite)
- Persona Prompting : "Tu es un juriste…", "Agis comme un expert du recrutement…"
- Structure logique de la tâche : objectif clair, consignes précises
- Exemples intégrés dans le prompt pour guider le modèle

##  Structuration et Design Contextuel
- **System Prompt** : définir la personnalité, le style, les règles de réponse du modèle
- **User Prompt** : exprimer une requête claire, contextualisée, actionnable
- Format de sortie imposé : JSON, tableau, markdown, YAML…
- Context Window Management : insérer les bons éléments dans la fenêtre contextuelle (historique, données, objectifs)

#  Étape 4 – Semantic Search & Retrieval-Augmented Generation (RAG)

La génération augmentée par récupération (RAG) permet de **combiner un modèle de langage avec vos propres données**. C’est l’étape clé pour construire des agents intelligents qui peuvent répondre sur des documents internes, des PDFs, des bases de connaissance ou des vidéos sous-titrées.

**L’objectif ici est d’acquérir :**
- La capacité à intégrer des sources externes au modèle
- La compréhension du pipeline RAG (embedding, retrieval, génération)
- La maîtrise des bibliothèques comme LangChain et LlamaIndex
- Le déploiement d’une chaîne complète : ingestion → question → réponse contextuelle

![RAG Illustration](img/rag.png)

*Source : [RAG from Scratch - LangChain AI](https://github.com/langchain-ai/rag-from-scratch)*

##  Objectif

Permettre à un LLM de s'appuyer sur des documents spécifiques ou des connaissances internes (PDF, vidéos, bases de données, contenus propriétaires) pour générer des réponses fiables, sourcées et à jour.

## Pipeline classique
1. Document brut → **chunking** (splitting en morceaux)
2. Chunk → vecteur via **embedding** (OpenAI, Hugging Face…)
3. Vecteurs stockés dans un **vector store** (FAISS, ChromaDB, Pinecone…)
4. Question utilisateur → convertie en vecteur → **recherche de passages proches**
5. Les passages sont injectés dans le prompt → **réponse LLM contextualisée**

## Concepts & Outils à maîtriser
- **FAISS** : recherche vectorielle locale rapide
- **ChromaDB** : vector store Python natif
- **Pinecone** : solution cloud évolutive
- **OpenAI Embeddings** ou **Hugging Face Sentence Transformers**
- **LangChain** : pipelines RAG simples et puissants
- **LlamaIndex** : ingestion intelligente de données hétérogènes

##  Cas d’usage typiques
- Résumer un PDF long en répondant à des questions ciblées
- Créer un moteur de recherche intelligent sur un corpus interne
- Fusionner des textes, vidéos, sous-titres pour générer des synthèses transmodales

###  Projets à réaliser :
- **Assistant IA sur base documentaire** : interroger une base de contrats, rapports ou politiques internes
- **Synthétiseur de réunions vidéo** : questionner l’enregistrement d’un call + transcription
- **Explorateur de réglementation** : interroger un corpus réglementaire à partir de cas d’usage

## Outils recommandés

- OpenAI / HuggingFace Embeddings
- ChromaDB / FAISS
- LangChain RAG
- LlamaIndex

### Aller plus loin : RAG Multimodal
- **RAG Vision** : utiliser GPT-4 Vision pour interroger des documents avec images, schémas ou captures PDF
- **RAG avec tableaux** : indexer des CSV ou extraits tabulaires via LangChain / Pandas + FAISS
- **RAG SQL** : combiner la recherche vectorielle avec des bases relationnelles (LangChain SQL + retriever hybride)

###  Ressources techniques RAG

- [LangChain RAG Examples](https://github.com/langchain-ai/langchain) – Exemples pratiques pour construire des chaînes de RAG avec divers backends.
- [LlamaIndex Recipes](https://github.com/run-llama/llama_index) – Démonstrations complètes avec ingestion, indexation et requêtage.

#  Étape 6 – Fine-tuning & Customisation des Modèles

Personnaliser un modèle de langage, que ce soit via le fine-tuning ou des méthodes plus légères comme l’adaptation par LoRA, permet de l’adapter à un contexte spécifique, un ton particulier ou un cas d’usage industriel. Cette étape est essentielle lorsqu’un prompt seul ne suffit plus.

**Objectif :**
Apprendre à ajuster, affiner et déployer un modèle (open-source ou propriétaire) pour un cas d’usage métier donné, en tenant compte des contraintes de données, d’infrastructure et de coût.

## Concepts clés
- Fine-tuning complet vs. Adapter tuning (LoRA, QLoRA, Prefix Tuning)
- Instruction tuning vs. domain adaptation
- RAG vs. fine-tuning : quand choisir l’un ou l’autre ?
- Importance du dataset : nettoyage, équilibre, supervision
- Évaluation : perplexité, exactitude, F1, hallucination

##  Types de fine-tuning
- **Instruction Tuning** : ajouter des exemples structurés pour affiner le comportement conversationnel
- **Domain Adaptation** : exposer le modèle à des données sectorielles (ex : finance, juridique…)
- **Low-Rank Adaptation (LoRA)** : méthode légère pour entraîner des poids résiduels sur petits GPU
- **QLoRA** : LoRA + quantization 4 bits pour faible consommation

##  Cas d’usage typiques
- Adapter Mistral ou LLaMA à un vocabulaire juridique
- Finetuner un modèle pour répondre à des formulaires internes (assurances, médical…)
- Créer un assistant RH spécialisé en recrutement tech

## Projets à réaliser
- **Fine-tuning LLaMA 2 avec LoRA** sur un corpus métier (PDF + instructions)
- **Créer un mini modèle finetuné sur dataset d’interviews RH** avec étiquettes de ton et d’émotion
- **Comparer un modèle avec prompt engineering seul vs. finetuné** sur un benchmark maison

## Outils & Frameworks
- Hugging Face Transformers + PEFT (LoRA)
- OpenLLM + BentoML pour le déploiement
- Axolotl, FastChat, QLoRA de Tim Dettmers
- Weights & Biases ou MLflow pour le suivi d’expériences
- Amazon SageMaker / Google Vertex AI pour entraîner à l’échelle

#  Étape 7 – Déploiement & Monitoring des LLMs

Une fois votre modèle (finetuné ou non) prêt, il est crucial de savoir le **déployer proprement, le superviser et l’optimiser en production**. Cette étape permet de passer d’un prototype local à une application fiable, sécurisée, scalable et observable.

**Objectif :**
Mettre en ligne un modèle (open-source ou via API), monitorer ses performances et intégrer des retours utilisateurs pour l’améliorer.

##  Concepts clés
- Modèle as a Service (MaaS), API REST vs gRPC
- Hosting local (GPU) vs cloud (SageMaker, Vertex AI, Replicate…)
- Serveurs de modèles : vLLM, Triton, Ray Serve
- Logging, versioning et monitoring des appels

## Méthodes de déploiement
- **API REST / FastAPI** : exposez votre modèle sous forme d’endpoint
- **Containerisation** : Docker + Kubernetes pour orchestrer à l’échelle
- **Hugging Face Inference Endpoint** : rapide, mais moins flexible
- **OpenLLM (BentoML)** : packaging + monitoring intégré

##  Outils à connaître
- **FastAPI / Uvicorn** : pour exposer un modèle via HTTP
- **Docker / Kubernetes** : containerisation et mise à l’échelle
- **OpenLLM** : déployer des modèles LoRA / finetunés avec suivi intégré
- **MLflow / Weights & Biases** : suivi d’expériences + logs de production
- **Langfuse / Traceloop / PromptLayer** : monitoring spécifique aux LLMs

## 🧪 Cas d’usage typiques
- Exposer un modèle finetuné via une API pour une app métier
- Surveiller les temps de réponse, les erreurs, les hallucinations
- Afficher les logs et feedbacks utilisateurs dans un tableau de bord

### 💡 Projets à réaliser

- **Déploiement sur Hugging Face + Langfuse** : suivez les réponses, prompts, erreurs, etc.
